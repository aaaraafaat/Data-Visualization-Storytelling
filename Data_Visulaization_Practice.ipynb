```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

# Set random seed for reproducibility
np.random.seed(42)

# Generate synthetic sales data
num_rows = 100

# 1. Transaction ID
transaction_ids = [f'TXN{str(i).zfill(5)}' for i in range(1000, 1000 + num_rows)]

# 2. Date (random dates in 2023)
start_date = datetime(2023, 1, 1)
end_date = datetime(2023, 12, 31)
dates = [start_date + timedelta(days=np.random.randint(0, 365)) for _ in range(num_rows)]

# 3. Product Category
categories = ['Electronics', 'Clothing', 'Home & Garden', 'Books', 'Sports', 'Beauty']
product_categories = np.random.choice(categories, num_rows, p=[0.3, 0.25, 0.2, 0.1, 0.1, 0.05])

# 4. Product Name (based on category)
def generate_product_name(category):
    if category == 'Electronics':
        return np.random.choice(['Smartphone', 'Laptop', 'Headphones', 'Tablet', 'Smartwatch'])
    elif category == 'Clothing':
        return np.random.choice(['T-Shirt', 'Jeans', 'Dress', 'Jacket', 'Shoes'])
    elif category == 'Home & Garden':
        return np.random.choice(['Chair', 'Table', 'Lamp', 'Plant', 'Decor'])
    elif category == 'Books':
        return np.random.choice(['Novel', 'Textbook', 'Cookbook', 'Biography', 'Children'])
    elif category == 'Sports':
        return np.random.choice(['Basketball', 'Yoga Mat', 'Running Shoes', 'Dumbbells', 'Bicycle'])
    else:
        return np.random.choice(['Perfume', 'Skincare Set', 'Makeup Kit', 'Hair Care', 'Body Lotion'])

product_names = [generate_product_name(cat) for cat in product_categories]

# 5. Quantity
quantities = np.random.randint(1, 6, num_rows)

# 6. Unit Price ($)
unit_prices = np.round(np.random.uniform(10, 500, num_rows), 2)

# 7. Total Amount ($)
total_amounts = np.round(quantities * unit_prices, 2)

# 8. Customer Segment
customer_segments = np.random.choice(['New', 'Returning', 'VIP'], num_rows, p=[0.4, 0.5, 0.1])

# 9. Payment Method
payment_methods = np.random.choice(['Credit Card', 'Debit Card', 'PayPal', 'Cash'], num_rows, p=[0.4, 0.3, 0.2, 0.1])

# 10. Region
regions = np.random.choice(['North', 'South', 'East', 'West'], num_rows, p=[0.3, 0.25, 0.25, 0.2])

# Create DataFrame
sales_data = pd.DataFrame({
    'Transaction_ID': transaction_ids,
    'Date': dates,
    'Product_Category': product_categories,
    'Product_Name': product_names,
    'Quantity': quantities,
    'Unit_Price': unit_prices,
    'Total_Amount': total_amounts,
    'Customer_Segment': customer_segments,
    'Payment_Method': payment_methods,
    'Region': regions
})

# Display the first 10 rows
print("Sales Dataset (First 10 rows):")
print(sales_data.head(10))

print(f"\nDataset Shape: {sales_data.shape}")
print(f"\nDataset Info:")
print(sales_data.info())

print(f"\nSummary Statistics:")
print(sales_data.describe())

```
