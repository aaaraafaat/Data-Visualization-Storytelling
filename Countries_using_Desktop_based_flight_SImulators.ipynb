```python
import plotly.graph_objects as go

# Data based on publicly available information on military VR/MR adoption or trials
countries = {
    'USA': {'lat': 38.9637, 'lon': -95.7129, 'text': 'Pioneered Pilot Training Next (PTN) program.'},
    'GBR': {'lat': 55.3781, 'lon': -3.4360, 'text': 'Partner in PTN; VR/AR use in RAF for jets and helicopters.'},
    'CAN': {'lat': 56.1304, 'lon': -106.3468, 'text': 'RCAF using VR to address training backlogs and enhance skills.'},
    'AUS': {'lat': -25.2744, 'lon': 133.7751, 'text': 'RAAF using VR/MR for medical and procedural training.'},
    'DNK': {'lat': 56.2639, 'lon': 9.5018, 'text': 'Royal Danish Airforce using VR for procedure training.'},
    'CHN': {'lat': 35.8617, 'lon': 104.1954, 'text': 'Extensive use of VR/AR in pilot training for advanced aircraft.'},
    'BGD': {'lat': 23.6850, 'lon': 90.3563, 'text': 'Proposed site for new in-house MR/VR training facility.'}
}

fig = go.Figure(data=go.Scattergeo(
    lon=[data['lon'] for data in countries.values()],
    lat=[data['lat'] for data in countries.values()],
    text=[f"{country}: {data['text']}" for country, data in countries.items()],
    mode='markers',
    marker=dict(
        size=10,
        color='blue',
        line_width=1,
        sizemode='diameter'
    )
))

fig.update_layout(
    title_text='Global Adoption of VR/MR in Military Pilot Training',
    geo=dict(
        scope='world',
        projection_type='natural earth',
        showland=True,
        landcolor='rgb(217, 217, 217)',
        subunitcolor="rgb(255, 255, 255)",
    )
)

fig.show()
```
